# Visualize a volume of inference

```python
import numpy as np
import h5py as h5
import matplotlib.pyplot as plt
```

Note before starting:

The inference files may not open if you ran the docker container as root. You may need to chown these files so that they belong to your username.

```python
num = 6
seg_vol = np.load(f"/home/ubuntu/code/ffn/results/karlupia_inf/karlupia_inference_2512_{num}/0/0/seg-0_0_0.npz")
image_vol = np.array(h5.File(f"/home/ubuntu/code/ffn/third_party/karlupia_inf1/karlupia_inference_2512_{num}.h5", mode='r')['raw'])
```

```python
seg = seg_vol['segmentation']
seg.shape
```

```python
frame = 40
plt.figure(1, figsize=(12,12))
plt.imshow(image_vol[frame,:,:], cmap='gray')
plt.imshow(seg[frame,:,:], cmap='tab20', alpha=0.2)
```
